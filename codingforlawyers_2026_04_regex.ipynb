# **Regex tutorial for beginners in Python**

A regular expression (usually called regex) is a way to search for patterns in text.

You can use regex to find things like:

+ email addresses
+ phone numbers
+ dates
+ words that start with a capital letter
+ numbers in a sentence

In Python, regex is handled with the built-in re module.



### **1. First idea: regex is pattern matching**

Imagine this text:

```python
text = "My phone number is 12345."
```

If you want to find the number inside the sentence, regex lets you describe the pattern “one or more digits”.



### **2. Import the regex module**

```python
import re
```

That is all you need to start.



### **3. Your first regex search**


```python
import re

text = "My phone number is 12345."
match = re.search(r"\d+", text)

print(match.group())
```


### **What this means**
+ re.search(...) looks for the first match
+ r"\d+" is the regex pattern
+ \d means: a digit (0 to 9)
+ + **means**: one or more
+ match.group() gives you the actual matched text

So \d+ means:

find one or more digits in a row


### **4. Why do we write r"..."?**

You will often see regex patterns written like this:


```python
r"\d+"
```

The **r** means raw string.

That helps Python treat backslashes correctly.
For regex in Python, you should usually write patterns as raw strings.

**Good:**


```python
r"\d+"
```

**Less safe:**

"\d+"

## **8. The most common regex functions in Python**
### **re.search()**

Finds the first match anywhere in the text.


```python
import re

text = "Order number: 48392"
match = re.search(r"\d+", text)

if match:
    print(match.group())
```

### **re.findall()**

Finds **all** matches and returns them as a list.


```python
import re

text = "I have 3 cats and 2 dogs."
matches = re.findall(r"\d+", text)

print(matches)
```

### **re.match()**

Checks only at the beginning of the string.


```python
import re

text = "Hello world"
match = re.match(r"Hello", text)

print(match.group())
```

This works because Hello is at the start.

### **re.sub()**

Replaces matches.

```python
import re

text = "My number is 12345."
new_text = re.sub(r"\d+", "[NUMBER]", text)

print(new_text)
```

### **re.split()**

Splits text using a regex pattern.

```python
import re

text = "apple,banana;orange grape"
parts = re.split(r"[,; ]+", text)

print(parts)
```

## **9. Capturing groups with ( )**

Groups let you extract parts of a match.

###Example: date

```python
import re

text = "Today is 19-04-2026"
match = re.search(r"(\d{2})-(\d{2})-(\d{4})", text)

if match:
    print(match.group(0))  # full match
    print(match.group(1))  # first group
    print(match.group(2))  # second group
    print(match.group(3))  # third group
```

So:

+ group(0) = whole match
+ group(1) = first part
+ group(2) = second part
+ group(3) = third part

### **10. Example patterns you can actually use**
**Find all numbers**

```python
import re

text = "The prices are 15, 29 and 300 euros."
print(re.findall(r"\d+", text))
```

#### **Find words starting with a capital letter**


```python
import re

text = "Anna lives in Berlin with Max."
print(re.findall(r"\b[A-Z][a-z]*\b", text))
```

###What this means
+ \b = word boundary
+ [A-Z] = one uppercase letter
+ [a-z]* = zero or more lowercase letters


#### **Find a simple email address**

```python
import re

text = "Write to me at test@example.com"
match = re.search(r"\w+@\w+\.\w+", text)

if match:
    print(match.group())
```

This is a simple email regex, not a perfect one, but it is good for learning.

#### **Find a German-style date like 19.04.2026**


```python
import re

text = "The deadline is 19.04.2026."
match = re.search(r"\d{2}\.\d{2}\.\d{4}", text)

if match:
    print(match.group())
```

Note: the dot must be written as \. because plain . means “any character”.

## **12. Word boundaries: \b**

A word boundary helps you match a whole word and not just part of a larger word.


```python
import re

text = "The cat scattered the catalog."
print(re.findall(r"\bcat\b", text))
```

Without \b, regex would also find cat inside catalog.

It does not. It means “any character”.

Use:

```python
r"\."
```

for an actual dot.


### **Mistake 3:** using search() when you need findall()

```python
r".+"
```

This matches a lot, often too much.

Try to be more specific.



## **17. A simple regex cheat sheet**


```python
\d      # digit
\w      # letter, digit, underscore
\s      # whitespace
.       # any character
+       # one or more
*       # zero or more
?       # zero or one
{3}     # exactly 3 times
{2,5}   # 2 to 5 times
[abc]   # one of a, b, c
[a-z]   # one lowercase letter
[^0-9]  # not a digit
^       # start of text
$       # end of text
\b      # word boundary
()      # capture group
|       # OR
```

## **23. Advanced techniques: groups**

### **Regex with groups**

```python
import re

text = "Today is 04-05-2026"
match = re.search(r"(\d{2})-(\d{2})-(\d{4})", text)

if match:
    day = match.group(1)
    month = match.group(2)
    year = match.group(3)

    print(day, month, year)
```

### **Regex with named groups**

```python
import re

text = "Today is 04-05-2026"
match = re.search(r"(?P<day>\d{2})-(?P<month>\d{2})-(?P<year>\d{4})", text)

if match:
    print(match.group("day"))
    print(match.group("month"))
    print(match.group("year"))
```
